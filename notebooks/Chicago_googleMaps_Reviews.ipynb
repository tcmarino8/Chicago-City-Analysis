```python
import requests

r = requests.get(
    "https://maps.googleapis.com/maps/api/place/textsearch/json",
    params={
        "query": "restaurants in Chicago",
        "key": API_KEY,
    },
)

data = r.json()

for place in data["results"][:10]:
    print(place["name"], place["place_id"])
```

```python
def get_reviews(place_id):
    r = requests.get(
        "https://maps.googleapis.com/maps/api/place/details/json",
        params={
            "place_id": place_id,
            "fields": "name,geometry,reviews",
            "key": API_KEY,
        },
    )
    return r.json()
```

```python
all_reviews = []

for place in data["results"]:
    placename, place_id = place["name"], place["place_id"]
    details = get_reviews(place_id)

    for review in details.get("result", {}).get("reviews", []):
        all_reviews.append({
            "restaurant": details["result"]["name"],
            "lat": details["result"]["geometry"]["location"]["lat"],
            "lng": details["result"]["geometry"]["location"]["lng"],
            "time": review["time"],
            "text": review["text"],
        })

latest = max(all_reviews, key=lambda r: r["time"])
print(latest)
```

```python
for i in range(5):
    print(all_reviews[i]['text'])
```

```python

# Create map centered on the average location
center_lat = sum(r["lat"] for r in all_reviews) / len(all_reviews)
center_lng = sum(r["lng"] for r in all_reviews) / len(all_reviews)

m = folium.Map(location=[center_lat, center_lng], zoom_start=13)

for review in all_reviews:
    review_date = datetime.fromtimestamp(review["time"]).strftime("%Y-%m-%d")

    popup_html = f"""
    <b>{review['restaurant']}</b><br>
    <b>Date:</b> {review_date}<br><br>
    {review['text']}
    """

    folium.Marker(
        location=[review["lat"], review["lng"]],
        popup=folium.Popup(popup_html, max_width=400),
        tooltip=review["restaurant"]
    ).add_to(m)

m.save("restaurant_reviews_map.html")
```

```python
from openai import OpenAI
import json

client = OpenAI()

def analyze_review(review_text):
    response = client.responses.create(
        model="gpt-5-mini",
        input=f"""
Analyze this restaurant review.

Return only valid JSON:

{{
  "overall": "positive|neutral|negative",
  "food": "positive|neutral|negative|not_mentioned",
  "service": "positive|neutral|negative|not_mentioned",
  "atmosphere": "positive|neutral|negative|not_mentioned",
  "value": "positive|neutral|negative|not_mentioned",
  "wait_time": "positive|neutral|negative|not_mentioned",
  "summary": "one sentence"
}}

Review:
{review_text}
"""
    )

    return json.loads(response.output_text)
```

```python
from collections import defaultdict, Counter

restaurant_scores = defaultdict(list)

for review in all_reviews:
    result = analyze_review(review["text"])
    restaurant_scores[review["restaurant"]].append(result)
```

```python
for restaurant, analyses in restaurant_scores.items():
    food_positive = sum(
        a["food"] == "positive"
        for a in analyses
    ) / len(analyses)

    service_negative = sum(
        a["service"] == "negative"
        for a in analyses
    ) / len(analyses)

    print(
        restaurant,
        food_positive,
        service_negative
    )
```
